```python
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
import missingno as msno
```

```python
sales = pd.read_csv("../DSF4E_Challenge1/customers_sales.csv", )
```

```python
sales["id"] = (
    sales.groupby(["customer_name", "customer_branch"], sort=False, dropna=False)
         .ngroup() + 1
)

sales.to_csv("../DSF4E_Challenge1/customers_sales_with_ids.csv", index=False)
```

```python
df = sales.drop(columns=["customer_name", "customer_branch"])
df = df[["id"] + [col for col in df.columns if col != "id"]]
df.to_csv("output.csv", index=False)
```
