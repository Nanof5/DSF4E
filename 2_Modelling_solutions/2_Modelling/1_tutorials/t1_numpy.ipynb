#### What is NumPy for?
NumPy is a library for numerical mathematics in Python. 
It provides efficient data structures and functions for performing mathematical operations on large datasets, particularly arrays and matrices.
NumPy is the foundation for many other libraries in Data Science, such as pandas, SciPy, and scikit-learn, and also plays a crucial role in Machine Learning and Deep Learning frameworks like TensorFlow and PyTorch.

#### Technical Background
While Python is a relatively intuitive programming language, it is very inefficient and slow compared to compiled languages like C or C++.
This is why NumPy uses C++ in the background, behind the provided Python interface, to perform mathematical operations much faster than pure Python code (see the optional speed comparison in section 7).  
At the core of NumPy are NumPy arrays, which allow for the performant handling of vectors, matrices, and tensors. 

The official NumPy documentation can be found here: [NumPy Documentation](https://numpy.org/doc/stable/)

#### Dimensions, Shape and Axes

Before creating arrays, let's build an intuition of what the dimensions of an array mean, using the laser cutting data of session 1 as an example:

<img align="center" width=800 src="../5_figures/array_dimensions.svg"/>

* A **1D array** is a vector, e.g., the cutting speeds of all parts: one value per part. A 1D array is neither a row nor a column vector; NumPy simply prints it in one line.
* A **2D array** is a matrix or a table: one row per part (sample) and one column per feature. This is exactly the structure of the data we use to fit models.
* A **3D array** stacks several 2D arrays, e.g., a set of grayscale images (image × pixel row × pixel column). Arrays with three or more dimensions are often called *tensors*; they come back in session 5 (Deep Learning).

Three terms describe the structure of an array:

|Term |Meaning |Example (2D table above) |
|---|---|---|
|`ndim` |Number of dimensions (axes) |`2` |
|`shape` |Number of elements along each axis |`(4, 3)`: 4 rows, 3 columns |
|`axis` |Index of a dimension: axis 0 runs down the rows, axis 1 across the columns |`axis=0`: along the parts |

Keep this picture in mind: many functions take an `axis` argument, and unexpected shapes are the most common source of errors when working with NumPy.

#### Accessing Elements

Elements of NumPy arrays can be accessed similarly to Python lists, using `[]` and indices.

```python
np_array = np.array([10, 20, 30, 40, 50])
# Accessing the third element (index 2)
print(np_array[2])
print(np_array[-1])
```

For multi-dimensional arrays, elements can be accessed using a comma-separated list of indices within the brackets:

```python
np_matrix = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
# Accessing a single element (row 1, column 2)
print(np_matrix[1, 2])  # Note: Indices start at 0
```

Using the colon `:` operator, we can access slices of the array.

**Predict first:** Which shape does the column `np_matrix[:, 1]` have: `(3,)`, `(1, 3)` or `(3, 1)`?

```python
np_matrix[0, :]  # Accessing all columns of row 0. Equivalent to np_matrix[0]
```

#### Boolean Indexing

As in pandas, elements can be selected with conditions. A comparison returns a *boolean mask* of the same shape as the array, which can be used as an index:

```python
np_array = np.array([10, 20, 30, 40, 50])
mask = np_array > 25  # Element-wise comparison
display(mask)
display(np_array[mask])  # Only the elements where the mask is True
```

Conditions are combined with `&` (and) and `|` (or). As in pandas, each condition must be put in parentheses:

```python
np_array[(np_array > 15) & (np_array < 45)]
```

#### Combining Arrays

`np.column_stack()` combines 1D arrays of the same length as the columns of a 2D array.

**Predict first:** Which shape does the result have?

```python
x0 = np.ones(3)
x1 = np.array([0.5, 1.5, 2.5])
x01 = np.column_stack([x0, x1])
display(x01)
print(x01.shape)  # One row per element, one column per input array
```

`np.concatenate()` joins arrays along an existing axis:

```python
np.concatenate([np_matrix, np_matrix], axis=0)  # Join along the rows: shape (6, 3)
```

## 3. Mathematical Operations <a id=math>
[Back to Overview](#ov)

Mathematical operations can now be performed quickly and easily on these NumPy arrays.
Simple arithmetic operations such as addition, subtraction, multiplication, and division can be applied directly to NumPy arrays using standard numeric values or other NumPy arrays of compatible shapes.

Let's see a few examples:

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
A * 2 # Multiply all elements by 2
```

```python
A + 2 # Add 2 to all elements
```

```python
A + A # Add two NumPy arrays element-wise
```

```python
A * A # Element-wise multiplication of two NumPy arrays (no matrix multiplication!)
```

#### Vector arithmetics

Using a numpy array of shape (1, n) and its transpose of shape (n, 1), we can calculate with row and column vectors.

Note that `*` is still an element-wise multiplication: multiplying a row vector of shape (1, 3) with a column vector of shape (3, 1) does **not** compute a dot product. Instead, both vectors are stretched to shape (3, 3) and multiplied element-wise, which results in the outer product of the two vectors. This behavior is called *broadcasting* and is explained in more detail below.

```python
a = np.array([[1,2,3]])
a * np.transpose(a)
```

#### Statistical functions

Analogously to pandas, many arithmetic and statistical functions are implemented in NumPy, such as minimum, maximum, sum, mean, median, standard deviation, etc.:

```python
print(a.min())  # Minimum of all elements
print(a.max())  # Maximum of all elements
print(a.sum())  # Sum of all elements
print(a.mean())  # Mean of all elements
print(a.std())  # Standard deviation of all elements

print(np.median(a))  # Median of all elements (note that this is not a method of the array object, but a NumPy function)
print(np.quantile(a, 0.25))  # 25th percentile of all elements (note that this is not a method of the array object, but a NumPy function)
print(np.mean(a))  # Mean of all elements (also available as a NumPy function)
```

#### Matrix operations

Using multidimensional arrays, we can also perform matrix operations such as matrix multiplication, summing all elements, or finding the minimum of each row or column:

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
A
```

```python
A.min()  # Minimum of all elements
```

Many functions take an `axis` argument, which specifies the axis **along** which the operation is performed.

**Predict first:** `A` has shape `(2, 3)`. How many values do `A.min(axis=0)` and `A.mean(axis=1)` return?

```python
A.min(axis=0)  # Minimum of each column
```

```python
A.mean(axis=1)  # Mean of each row
```

The axis along which we aggregate disappears in the result (compare the figure in section 2):
* `axis=0` runs down the rows, so we get one value per column: shape `(3,)`.
* `axis=1` runs across the columns, so we get one value per row: shape `(2,)`.

For the laser data with one row per part, `axis=0` therefore computes a statistic per feature, e.g., the mean of each column.

The transpose of a matrix can be obtained using the `T` attribute (or, equivalently, the `transpose()` method):

```python
A.T # Transpose using the T attribute
```

The matrix multiplication is performed with the `@` operator (or, equivalently, the `dot()` method):

```python
A @ A.T  # Matrix multiplication using the @ operator
```

#### Broadcasting

*Broadcasting* in NumPy is a powerful concept that allows arithmetic operations to be performed on arrays with different, but compatible, shapes. 

For example, a 2x3 matrix can be added to a 1x3 matrix, which simply applies the addition of the 1x3 matrix to both rows of the 2x3 matrix:

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[1, 2, 3]])
A + b # Broadcasting: Vector b is added to each row of matrix A
```

Beware, this also means that many operations that you might expect to raise an error will actually work due to broadcasting.
This can sometimes lead to unexpected results if you are not aware of this behavior.
It is a good idea to always check the shapes of the arrays involved in operations to ensure they are as intended.

Applying an arithmetic operation with a scalar, as shown earlier (e.g., `A * 2`), is also a form of broadcasting.

Due to its efficiency, broadcasting is widely used, especially in the fields of Big Data and Machine Learning, and it is a frequently used concept in pandas.

## 4. Solving Linear Equation Systems <a id=linalg>
[Back to Overview](#ov)

Many problems in engineering and data science lead to a system of linear equations $A \mathbf{x} = \mathbf{b}$, e.g., the closed-form solution of linear regression.
The submodule `np.linalg` provides functions for linear algebra.

As an example, we solve the following system with two equations and two unknowns:

$$
2 x_1 + x_2 = 5 \\
x_1 + 3 x_2 = 10
$$

```python
A = np.array([[2, 1], [1, 3]])  # Coefficients of the equations
b = np.array([5, 10])  # Right-hand side
x = np.linalg.solve(A, b)
x
```

We can check the solution by inserting it into the equations. As floating point numbers are subject to rounding errors, we compare with `np.allclose()` instead of `==`:

```python
display(A @ x)
np.allclose(A @ x, b)
```

Alternatively, the system can be solved with the inverse matrix: $\mathbf{x} = A^{-1} \mathbf{b}$.

```python
A_inv = np.linalg.inv(A)
A_inv @ b
```

Both ways give the same result. In practice, `np.linalg.solve()` is preferred: it is faster and numerically more accurate, as the inverse is never computed explicitly.

If the matrix $A$ is *singular*, i.e., some of its rows or columns are linearly dependent, the system has no unique solution and NumPy raises an error.
In linear regression, this happens if two features carry exactly the same information (perfect collinearity, compare Step 5 of the EDA):

```python
A_singular = np.array([[1, 2], [2, 4]])  # The second row is 2 times the first row
try:
    np.linalg.solve(A_singular, b)
except np.linalg.LinAlgError as error:
    print('LinAlgError:', error)
```

## 5. Data Types <a id=datatypes>
[Back to Overview](#ov)

Since NumPy uses C in the background, all elements of an array have the same, fixed data type (`dtype`). The data type defines how much memory each element needs and which values it can represent.

In NumPy, there are 4 different data types for integers alone, each with a different value range:

|Data Type |Description |Size (Bytes)|Value Range |
|---|---|---|---|
|`np.int8` |8-bit integer |1 |-128 to 127 |
|`np.int16` |16-bit integer |2 |-32768 to 32767 |
|`np.int32` |32-bit integer |4 |-2147483648 to 2147483647 |
|`np.int64` |64-bit integer |8 |-9223372036854775808 to 9223372036854775807 |

Additionally, there are other data types in NumPy, e.g., positive integers (`uint`) and floating point numbers (`float`) of different sizes.

To specify the data type of a NumPy array, the `dtype` parameter can be used when creating the array:

```python
np.array([1, 2, 3], dtype=np.int8)  # Create an array of 8-bit integers
```

If no data type is specified, NumPy infers it from the input: integers become `np.int64` (64-bit integer) and floating point numbers become `np.float64` (64-bit floating point number).
Functions like `np.zeros()` and `np.ones()` create arrays of type `np.float64` by default.

```python
np_array = np.array([1, 2, 3])  # Integer input: default data type is np.int64
display(np_array) # Display the array (dtype is omitted in default case when displaying)
display(np_array.dtype) # Check the data type of the array
display(np.array([1.0, 2.0, 3.0]).dtype)  # Floating point input: default data type is np.float64
```

The data type within a NumPy array can be changed using the `astype()` method:

```python
np_array = np.array([1, 2, 3], dtype=np.float32)
display(np_array)
np_array_double = np_array.astype(np.double)  # Convert to double
display(np_array_double) # Note that this is equivalent to np.float64 and is the default number type in NumPy
np_array_int8 = np_array.astype(np.int8)  # Convert to int8
display(np_array_int8)
```

#### Pitfall: Shapes `(n,)` and `(n, 1)`

A 1D array of shape `(n,)` and a 2D column vector of shape `(n, 1)` look similar, but behave differently in combination. Due to broadcasting, subtracting one from the other does **not** raise an error, but creates an `(n, n)` matrix:

```python
y = laser['Gth_micro'].to_numpy()  # Shape (n,)
y_col = y.reshape(-1, 1)  # Shape (n, 1); -1 means: infer this dimension from the number of elements
print(y.shape, y_col.shape)
print((y - y_col).shape)  # Not n differences, but an n x n matrix!
```

When computing errors between true and predicted values, this mistake silently leads to wrong results. Check the shapes of your arrays regularly, e.g., with `assert y_pred.shape == y.shape`.

#### Comparison: NumPy vs. Pure Python

To demonstrate the speed of NumPy, we first create a long list of random values and a corresponding NumPy array from that list:

```python
import random

# Create a Python list with 1,000,000 random values between 0 and 1
python_list = [random.random() for _ in range(1000000)]

numpy_array = np.array(python_list)
```

While we can investigate the length of the Python list using `len()`, the dimensions of a NumPy array are given by its `shape` attribute (see section 2).

```python
# Length of the Python list
display(len(python_list))
# Length and shape of the NumPy array
display(len(numpy_array)) # Only helpful for one-dimensional arrays
display(numpy_array.shape) # Preferred way to get dimensions of NumPy arrays
```

Now, we will sum all the values in the list and measure the time it takes. First in pure Python, then with NumPy.

For the Python list, we will have to implement a simple loop ourselves:

```python
%%timeit # Jupyter Notebook-specific feature to measure the execution time of a command. Note that the code cell will be executed multiple times to get a reliable measurement.
sum = 0 
for i in python_list:
    sum += i
```

The efficient numpy implementation for summing the elements of an array is called `sum()`:

```python
%%timeit
numpy_sum = np.sum(numpy_array)
```

As we can see, NumPy takes only a fraction of the time for this task compared to pure Python.

349 μs = 0.349 ms vs. 19 ms is a speedup of more than 50 times!

(Note that the exact times may vary depending on your hardware and current system load, but the difference in speed should be significant in any case.)
