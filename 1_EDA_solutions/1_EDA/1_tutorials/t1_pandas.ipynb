Important properties of DataFrames:
* A DataFrame is 2-dimensional, such that it has rows and columns and can contain different data types such as int, float, or strings.

### Task 4: Filtering rows

<img align="center" width=600 src="../5_figures/data_frame_filter_row.png"/>

It is also interesting to only look at certain rows of a DataFrame.
Perhaps you only want to look at those students whose grade is better than or equal to 1.3.
You can achieve this via filtering by describing the condition within the square brackets ``[]``.
The condition results in one truth value (``True`` or ``False``) per row, and only the rows with ``True`` are kept:

```python
best_students = students[students["exam_grade"] <= 1.3]
best_students
```

To implement multiple conditions, you can implement the conditions inside parentheses and link them to each other using logical operators.

**Important**: To concatenate conditions in pandas, use the **&** character instead of the **and** operator, and the **|** character (vertical bar) instead of the **or** operator.

*Why?* A condition such as ``students["exam_grade"] <= 1.3`` does not result in a single ``True`` or ``False``, but in a ``Series`` with one truth value per row.
Python's ``and`` and ``or`` need a single truth value, so they raise an error for a ``Series``.
pandas defines ``&`` (and), ``|`` (or) and ``~`` (not) to combine such ``Series`` **row by row**, which results again in one truth value per row.
As ``&`` and ``|`` are evaluated before comparisons like ``<=`` or ``==``, each condition must be put in parentheses.

```python
best_me_students = students[
    (students["exam_grade"] <= 1.3) & 
    (students["course_of_study"] == "ME")
]

best_me_students
```

```python
students[
    (students["exam_grade"] <= 1.3) | 
    (students["course_of_study"] == "EE")
]
```

### Task 5: Selecting Rows and Columns with ``.loc`` and ``.iloc``

So far, we selected columns with ``[]`` and rows with filter conditions. With ``.loc`` and ``.iloc``, we can select rows **and** columns at the same time:

* ``df.loc[rows, columns]`` selects by **labels**, i.e., row index labels and column names.
* ``df.iloc[rows, columns]`` selects by **integer positions** (0, 1, 2, ...), just like indexing a Python list.

<img align="center" width=600 src="../5_figures/data_frame_loc_iloc.png"/>

Remember that we removed the row with index label ``4`` from ``students`` earlier. Therefore, row labels and row positions no longer coincide:

```python
display(students)
```

```python
# Label-based: the row with index label 5, column "surname"
students.loc[5, "surname"]
```

```python
# Position-based: the 5th row (position 4) and the 2nd column (position 1) - the same entry!
students.iloc[4, 1]
```

Both accept lists and slices for rows and columns.

**Important**: slices with ``.loc`` **include** the end label, whereas slices with ``.iloc`` **exclude** the end position (as for Python lists).

```python
# Rows with labels 0 to 2 (including 2), two columns selected by name
students.loc[0:2, ["surname", "exam_grade"]]
```

```python
# Rows at positions 0 to 1 (excluding 2), columns at positions 1 and 3
students.iloc[0:2, [1, 3]]
```

Instead of labels, ``.loc`` also accepts a **filter condition** for the rows. This is the recommended way to **change values** in selected rows.

As an example, assume that the exam grade of the student with matriculation number ``54365`` was recorded incorrectly and should be ``3.7``. We correct it in a copy of the DataFrame:

```python
students_corrected = students.copy()
students_corrected.loc[students_corrected["matriculation_number"] == 54365, "exam_grade"] = 3.7
students_corrected
```

**Watch out:** Do not use two consecutive brackets (so-called *chained indexing*) to change values, e.g., ``students[students["matriculation_number"] == 54365]["exam_grade"] = 3.7``.
The first bracket creates a temporary copy, and only this copy is changed - the original DataFrame stays unchanged (pandas shows a warning in this case).
Always use ``.loc[rows, column]`` to change values.

### Task 6: Creating Categorical Columns via Mapping

Categorical values are often stored as **numeric codes** in tabular data (e.g., ``1`` for ``"female"``, ``2`` for ``"male"``). Before analysis, we usually want to convert these codes into readable labels.

The pandas ``Series.map()`` method translates each value through a **dictionary**. To avoid accidentally overwriting the original data, we assign the result to a **new column** and keep the original one.

Let's first create a small example ``DataFrame``:

```python
students2 = pd.DataFrame({
    "name": ["Anna", "Ben", "Cleo", "Dan", "Eva"],
    "gender_code": [1, 2, 1, 2, 1],
    "grade": [1.0, 2.3, 1.7, 3.0, 2.0],
})
display(students2)
```

The column ``gender_code`` contains the values ``1`` and ``2``. Let's decode these values with ``map()`` and store the labels in a **new** column ``gender``:

```python
gender_mapping = {1: "female", 2: "male"}

students2["gender"] = students2["gender_code"].map(gender_mapping)
display(students2)
```

Notice that the original column ``gender_code`` is **still there**. This is safer than overwriting it, because we can then compare or revert.

**a)**  Use ``map()`` to add a **new** column ``grade_label`` to ``students2``, which labels the ``grade`` column using the provided dictionary. Use ``display`` to show the resulting ``DataFrame``.

### Task 7: Grouping and Aggregating Data

Often, we want to compute statistics for **groups** of rows, e.g., the average exam grade per course of study.
The method ``groupby()`` implements this in three steps (*split - apply - combine*):

1. **Split** the rows into groups according to the values of one (or more) columns,
2. **apply** a function (e.g., ``mean()``) to each group, and
3. **combine** the results into a new ``Series`` or ``DataFrame``, whose index consists of the group values.

```python
students.groupby("course_of_study")["exam_grade"].mean()
```

With ``agg()``, we can compute several statistics at once by passing a list of their names:

```python
students.groupby("course_of_study")["exam_grade"].agg(["mean", "min", "max", "count"])
```

When grouping by **several** columns, the result has a combined index with one level per column.
The method ``unstack()`` turns the inner index level into columns, which results in a clearly arranged table (``NaN`` means that there is no row for this combination):

```python
grade_counts = students2.groupby(["gender", "grade"])["name"].count()
display(grade_counts)
display(grade_counts.unstack())
```

### Task 8: Combining DataFrames with ``merge()``

Information is often spread over several tables. For example, our ``students`` table only contains the abbreviation of the course of study, while the full names are stored in a separate table ``courses``:

```python
courses = pd.DataFrame({
    "course_of_study": ["IE", "ME", "CS"],
    "course_name": ["Industrial Engineering", "Mechanical Engineering", "Computer Science"],
})
display(courses)
```

The function ``pd.merge(left, right, on="key_column", how=...)`` combines the rows of two DataFrames that have the same value in the key column.
The parameter ``how`` determines what happens with rows that have **no** matching partner in the other table:

| ``how`` | Result contains |
|---|---|
| ``"inner"`` (default) | only rows whose key exists in **both** tables |
| ``"left"`` | **all** rows of the left table; missing information from the right table becomes ``NaN`` |
| ``"right"`` | **all** rows of the right table |
| ``"outer"`` | **all** rows of **both** tables |

In our example, there is no course name for ``EE``, and there is no student of ``CS``:

```python
for how in ["inner", "left", "outer"]:
    merged = pd.merge(students, courses, on="course_of_study", how=how)
    print(f'how="{how}": {merged.shape[0]} rows')
    display(merged)
```

Note that the inner join **lost** the student studying ``EE``! It is good practice to compare the number of rows before and after merging.

*Hint:* If the key columns have different names in the two tables, rename one of them first (``rename()``) or use the parameters ``left_on`` and ``right_on`` instead of ``on``.
