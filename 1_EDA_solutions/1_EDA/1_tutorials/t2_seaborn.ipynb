### Your First Chart

We start with a simple chart, the **Scatter Plot**.
A scatter plot typically uses the values of two variables (columns), where the coordinates of the points in the chart correspond to the value pairs of the two variables in the available data.

The position of a point on the horizontal axis (**x-axis**) shows the value of one variable, and its position on the vertical axis (**y-axis**) shows the value of the other variable. Scatter plots are particularly useful for visually examining the relationship (**correlation**) between two variables.

Since Seaborn can work directly with Pandas DataFrames, we first load the tabular example dataset `yearly_flights.csv` into a DataFrame.
This dataset contains the total number of air passengers per year between 1949 and 1960.

```python
yearly_flights = pd.read_csv("../4_data/yearly_flights.csv") # This CSV file is located in this chapter's subfolder 4_data.

yearly_flights.head()
```

We can create a scatter plot by using the function `sns.scatterplot()`.

The function expects a DataFrame in the argument `data`.
Furthermore, two string arguments `x` and `y` are passed, which specify the names of the columns to be visualized on the respective axes.

We pass our DataFrame in the argument `data`,
as well as the column names `year` and `passengers` as the columns to be visualized via the two string arguments `x` and `y`.

```python
sns.scatterplot(data=yearly_flights, x="year", y="passengers")
```

Here, one can directly see a positive relationship between the two variables: the number of passengers increases every year.
The yearly increase even grows over time, so the relationship is not exactly linear.\
Note that both the plotted markings on the y-axis ($2000$, $3000$, etc.) and those on the x-axis were set automatically and are sorted in ascending order.
These markings are often also called "ticks".

### Styling

The previous graph already looks good!
We could, for example, add a grid to the background of the previous chart to improve **readability**.

Seaborn offers the **`set_style()`** function as a way to generate optimized graphics very easily and quickly, in which, for instance, colors, fonts, and plot elements are already coordinated.
Without calling this function, the default style of matplotlib is used.
You can try out the function with the parameters **`white`**, **`darkgrid`**, **`ticks`**, **`whitegrid`**, or **`dark`**.
After changing the Seaborn style, simply generate the chart above again and pay attention to the changing appearance.

You can find more on styling charts [here.](https://seaborn.pydata.org/tutorial/aesthetics.html)

```python
sns.set_style('darkgrid')
```

Now we draw the same chart again to see how it looks with the new style.

```python
sns.scatterplot(data=yearly_flights, x="year", y="passengers")
```

You can create many types of charts in Seaborn. We will only show the most important ones here.
Creating other chart types works very similarly and can be looked up in the documentation linked above.

### Line Chart
Line charts are similar to scatter plots; the values of two variables are also displayed on the **x-** and **y-axis**.
However, the data points are connected by straight lines, ordered by their x-values.
If there are several data points with the same x-value, `lineplot()` draws their mean together with a shaded confidence interval.

A line chart can be created with the function **`sns.lineplot()`**, which accepts the same arguments as the scatter plot.

```python
sns.lineplot(data=yearly_flights, x="year", y="passengers")
```

### Count Plot

Count plots show how often the different categories (values) of a **categorical** variable occur and can thus help to visualize and understand the **distribution** of this variable.
The categories are displayed, for example, along the **x-axis**, and the number of occurrences of each category is indicated by the height of the bar at the corresponding position.

For **numerical** variables with many different values (e.g., age or temperature), a **histogram** (`sns.histplot()`) is used instead, which groups the values into intervals (*bins*) and counts the values per interval.

First, we read in a tabular dataset about the passengers of the Titanic ship.

```python
titanic = pd.read_csv("../4_data/titanic.csv")

titanic.head(10)
```

The dataset contains information about the ship's passengers, such as age, gender, ticket class, and whether a passenger survived.

We will now draw a chart that represents the count of surviving passengers.
For this, we use the function for creating count plots, `sns.countplot()`, and pass it the DataFrame, as well as the name of the column to be considered.

```python
sns.countplot(data=titanic, x="Survived")
```

The chart shows us that the majority of passengers did not survive (indicated by the variable value **0** in the dataset).

The frequency of other, non-numeric columns can also be considered, e.g., the gender of the passengers (values 'male', 'female'):

```python
sns.countplot(data=titanic, x="Sex")
```

## 3. Displaying Subset Data <a id="3-displaying-subset-data"></a>

[Back to top](#contents)


In the last charts, we plotted the number of surviving passengers and their gender in isolation from each other.
However, we can combine both charts by displaying **subset data** and representing the count of survivors, broken down by their **gender**, in different colors.

This can be easily implemented with the `countplot()` function using the **`hue`** argument:

```python
sns.countplot(data=titanic, x="Survived", hue="Sex")
```

Here, the absolute frequencies of the values in the `Survived` column are displayed separately according to the existing categories in the `Sex` column.

The **`hue`** argument is also available when creating many other chart types in Seaborn.

Similar to the example seen before, we now read a CSV file into Pandas that contains the absolute number of air passengers by month and year.

```python
monthly_flights = pd.read_csv("../4_data/monthly_flights.csv")

monthly_flights.head()
```

We plot the total number of passengers per month in a line chart.
We display the year on the x-axis and the passenger count on the y-axis and use the 'hue' argument to distinguish the data of the different months from each other.

```python
sns.lineplot(data=monthly_flights, x="year", y="passengers", hue="month")
```

This chart tells us a lot:

* The number of passengers increases over the years for every month
* The number of passengers in July and August is very similar
* Generally, more people fly in the middle of a year (May-September) than at the beginning or the end

Seaborn offers a variety of different plot types, which you can find in the official seaborn documentation.

The API for all plot types is very similar.
In most cases, you can
* Pass the entire DataFrame as named argument `data` (or the plot-functions first argument)
* Specify column names via named parameters `x` (and `y`)
* Add a coloured dimension by passing a column name to the named argument `hue`

## 4. Titles, Customizing and Saving Charts <a id="4-titles-customizing-and-saving-charts"></a>

[Back to top](#contents)

### Setting the Title

It is common and helpful to add a **title** to created charts that summarizes the content of the diagram.

Seaborn plotting functions return the chart (a matplotlib *axes* object), which we store in a variable, e.g., `plot`.
Using its method **`set_title()`**, a title can be added to the chart. Similarly, `set_xlabel()` and `set_ylabel()` change the axis labels:

```python
# The same chart about the number of passengers
plot = sns.lineplot(data=monthly_flights, x="year", y="passengers", hue="month")
plot.set_title("Number of Passengers per Month between 1949 and 1960")
```

### Customizing and Saving Charts

Seaborn creates its charts with the library **`matplotlib`**, which is usually imported as `plt`.
Via `plt`, the most recently created chart can be modified further, e.g., its title, the labels, ticks, and ranges of the x- and y-axis, or the legend.

The chart can also be saved to a file to be used later, e.g., in presentations or scientific papers: `plt.savefig()` writes the most recently created chart to the given file (here "passengers_lineplot.png" in this notebook's folder).
Finally, `plt.show()` displays the chart. Call `plt.savefig()` **before** `plt.show()`.

```python
import matplotlib.pyplot as plt

plot = sns.lineplot(data=monthly_flights, x="year", y="passengers", hue="month")
# Note different way to set the title
plt.title("Number of Passengers per Month between 1949 and 1960")
# Plot customization via plt
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=45)
plt.xlim(1948, 1961)
plt.legend(title="Month", loc="lower left")
# Save the chart to a file and show it
plt.savefig("passengers_lineplot.png")
plt.show()
```

### Several Charts in One Figure

`plt.subplots(nrows, ncols)` creates a figure with a grid of charts and returns the figure and an array of *axes* objects (one per chart).
Pass the axes in which a chart should be drawn to the seaborn function via the parameter `ax`.
With several rows **and** columns, `axes` is two-dimensional, e.g., `axes[0][1]` is the chart in the first row and second column.

Note that `plt` functions like `plt.xticks()` only change the most recently drawn chart - use the methods of the axes objects instead, e.g., `axes[0].set_title()`.

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))
sns.lineplot(data=yearly_flights, x="year", y="passengers", ax=axes[0])
axes[0].set_title("Passengers per Year")
sns.countplot(data=titanic, x="Survived", ax=axes[1])
axes[1].set_title("Survivors of the Titanic")
plt.suptitle("Two Charts in One Figure")
plt.tight_layout()
plt.show()
```
